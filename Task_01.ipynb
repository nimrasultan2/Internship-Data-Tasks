```python
import pandas as pd
import numpy as np


df = pd.read_csv ('/content/sample_data/mnist_test.csv')

print ('Inpecting Data\n')
print (f'Shape: {df.shape}\n')
print (f'Info: {df.info()}\n')
print (f'Columns: {df.columns}\n')
print (f'Head: \n {df.head()}\n')

df = df.dropna()
df = df.drop_duplicates()

print ('\nConverting into NumPy array')
arr = df['0.105'].values
new_arr = arr+10
print (new_arr)

mean = np.mean(new_arr)
median = np.median(new_arr)
std = np.std(new_arr)

print(f'\nBasic statistics')
print(f'Mean: {mean}')
print(f'Median: {median}')
print(f'Std: {std}')

```
