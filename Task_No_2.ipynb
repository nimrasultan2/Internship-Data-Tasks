```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = sns.load_dataset('iris')

#display data structure
print("\nShape of dataset:", df.shape)
print("\nColumns:", df.columns)
print("\nFirst 5 rows:")
print(df.head())

plt.figure(figsize=(8, 5))
#scatterplot to see if variables are correlated
sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species')
plt.title('\nScatter Plot:  Sepal Length vs. Sepal Width')
plt.show()

plt.figure(figsize=(8, 5))
#histplot to show how frequently data appears
sns.histplot(df['petal_length'], kde=True, color='skyblue')
plt.title('\nHistogram: Distribution of Petal Length')
plt.show()

plt.figure(figsize=(8, 5))
#boxplot to detect spread of values
sns.boxplot(x='species', y='petal_width', data=df)
plt.title('\nBox Plot: Petal Width by Species')
plt.show()





```
